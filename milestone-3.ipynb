```python
import torchaudio
waveform = torch.zeros(1, 16000)  # 1-second audio at 16kHz
mel = torchaudio.transforms.MelSpectrogram(sample_rate=16000, n_fft=400, hop_length=160, n_mels=64)(waveform)
print(mel.shape)
```

```python
import torch.nn as nn
x = torch.zeros(1, 1, 64, 64)
out = nn.MaxPool2d(2, 2)(nn.Conv2d(1, 1, kernel_size=3, stride=1, padding=1)(x))
print(out.shape[-2:])  
```
